# Árboles en Python

Los árboles son una estructura de datos fundamental en ciencias de la computación y programación. Un árbol consiste en un conjunto de nodos conectados por bordes, con un nodo raíz desde el cual se derivan los demás nodos. Cada nodo puede tener cero o más nodos hijos.

En esta guía, exploraremos cómo implementar y trabajar con árboles en Python.

## Definición de un Nodo

Primero, definiremos la clase `Nodo` que representará cada nodo en nuestro árbol.


```python
class Nodo:
    def __init__(self, key):
        self.izq = None
        self.der = None
        self.val = key
```

## Creación de un Árbol

A continuación, crearemos un árbol simple para ilustrar cómo se pueden conectar los nodos.


```python
# Crear el nodo raíz
raiz = Nodo(1)

# Añadir nodos hijos
raiz.izq = Nodo(2)
raiz.der = Nodo(3)

# Añadir nodos hijos al nodo izquierdo del raíz
raiz.izq.izq = Nodo(4)
raiz.izq.der = Nodo(5)

# Añadir nodos hijos al nodo derecho del raíz
raiz.der.izq = Nodo(6)
raiz.der.der = Nodo(7)

# Añadir nodos hijos al nodo hijo derecho del hijo izquierdo del raíz
raiz.izq.der.izq = Nodo(8)
raiz.izq.der.der = Nodo(9)

# Añadir nodos hijos al nodo hijo derecho del hijo derecho del raíz
raiz.der.der.izq = Nodo(10)
raiz.der.der.der = Nodo(11)
```

En el codigo anterior se creo el árbol de ejemplo inicial, en principio cargandolo por _niveles_. Veremos mas adelante otras formas de recorrer un árbol.

## Travesía de un Árbol

Existen diferentes maneras de recorrer un árbol. Aquí mostraremos tres métodos comunes: preorden, inorden y postorden.


### Travesía en Preorden

En la travesía en preorden, visitamos primero la raíz, luego el subárbol izquierdo, y finalmente el subárbol derecho.


```python
def print_preorder(raiz):
    if raiz:
        print(raiz.val, end=" ")
        print_preorder(raiz.izq)
        print_preorder(raiz.der)
        
print_preorder(raiz)  # Salida esperada: 1 2 4 5 8 9 3 6 7 10 11 
```

### Travesía en Inorden

En la travesía en inorden, visitamos primero el subárbol izquierdo, luego la raíz, y finalmente el subárbol derecho.


```python
def print_inorder(raiz):
    if raiz:
        print_inorder(raiz.izq)
        print(raiz.val, end=" ")
        print_inorder(raiz.der)
        
print_inorder(raiz)  # Salida esperada: 4 2 8 5 9 1 6 3 10 7 11
```

### Travesía en Postorden

En la travesía en postorden, visitamos primero el subárbol izquierdo, luego el subárbol derecho, y finalmente la raíz.


```python
def print_postorder(raiz):
    if raiz:
        print_postorder(raiz.izq)
        print_postorder(raiz.der)
        print(raiz.val, end=" ")
        
print_postorder(raiz)  # Salida esperada: 4 8 9 5 2 6 10 11 7 3 1 
```

```python
#Desafío 78

class Nodo:
    def __init__(self, key):
        self.izq = None
        self.der = None
        self.val = key

raiz = Nodo(10)
raiz.izq = Nodo(20)
raiz.der = Nodo(5)
raiz.izq.izq = Nodo(30)
raiz.izq.der = Nodo(15)

def maximo_postorden(raiz):
    if raiz is None:
        return None

    max_izq = maximo_postorden(raiz.izq)
    max_der = maximo_postorden(raiz.der)

    mayor = raiz.val

    if max_izq is not None and max_izq > mayor:
        mayor = max_izq

    if max_der is not None and max_der > mayor:
        mayor = max_der

    return mayor

resultado = maximo_postorden(raiz)
print("El valor máximo es:", resultado)
```

```python
#Desafío 79

class Nodo:
    def __init__(self, key):
        self.izq = None
        self.der = None
        self.val = key

def insertar(raiz, valor):
    if raiz is None:
        return Nodo(valor)

    if valor < raiz.val:
        raiz.izq = insertar(raiz.izq, valor)
    elif valor > raiz.val:
        raiz.der = insertar(raiz.der, valor)

    return raiz

def buscar(raiz, valor):
    if raiz is None:
        return False

    if raiz.val == valor:
        return True

    if valor < raiz.val:
        return buscar(raiz.izq, valor)
    else:
        return buscar(raiz.der, valor)

numeros = [10, 5, 15, 3, 7, 12, 20]

raiz = None

for numero in numeros:
    raiz = insertar(raiz, numero)

valor = int(input("Ingrese el número que desea buscar: "))

if buscar(raiz, valor):
    print("El número existe en el árbol.")
else:
    print("El número no existe en el árbol.")
```

```python
#Desafío 80

import re

class Nodo:
    def __init__(self, valor):
        self.val = valor
        self.izq = None
        self.der = None

def construir_arbol(expresion):
    operadores = []
    nodos = []
    prioridad = {"+": 1, "-": 1, "*": 2, "/": 2}

    def crear_nodo():
        operador = operadores.pop()
        derecho = nodos.pop()
        izquierdo = nodos.pop()

        nuevo = Nodo(operador)
        nuevo.izq = izquierdo
        nuevo.der = derecho
        nodos.append(nuevo)

    elementos = re.findall(r"\d+|[()+*/-]", expresion)

    for elemento in elementos:
        if elemento.isdigit():
            nodos.append(Nodo(int(elemento)))

        elif elemento == "(":
            operadores.append(elemento)

        elif elemento == ")":
            while operadores[-1] != "(":
                crear_nodo()
            operadores.pop()

        else:
            while (operadores and operadores[-1] != "("
                   and prioridad[operadores[-1]] >= prioridad[elemento]):
                crear_nodo()
            operadores.append(elemento)

    while operadores:
        crear_nodo()

    return nodos[0]

def evaluar(raiz):
    if raiz.izq is None and raiz.der is None:
        return raiz.val

    izquierda = evaluar(raiz.izq)
    derecha = evaluar(raiz.der)

    if raiz.val == "+":
        return izquierda + derecha
    elif raiz.val == "-":
        return izquierda - derecha
    elif raiz.val == "*":
        return izquierda * derecha
    elif raiz.val == "/":
        return izquierda / derecha

expresion = "5 + 3 * 4"

arbol = construir_arbol(expresion)
resultado = evaluar(arbol)

print("Expresión:", expresion)
print("Resultado:", resultado)
```
